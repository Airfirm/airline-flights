```python
import pandas as pd
import numpy as np

from pathlib import Path

print("Pandas version:", pd.__version__)
print("NumPy version:", np.__version__)
```

```python
# Define the project directory
project_root = Path(r"C:\Repo\airline-flights")

# Define data directories
raw_path = project_root / "data" / "raw"
processed_path = project_root / "data" / "processed"

# Ensure processed directory exists
processed_path.mkdir(parents=True, exist_ok=True)

# Find all monthly CSV files
csv_files = sorted(raw_path.glob("flights_2025_*.csv"))

print("Number of datasets found:", len(csv_files))

for file in csv_files:
    print(file.name)
```

```python
# Read column names without loading all records
monthly_columns = {}

for file in csv_files:

    columns = pd.read_csv(
        file,
        nrows=0
    ).columns.tolist()

    monthly_columns[file.name] = columns

# Compare each dataset with January
reference_columns = monthly_columns[csv_files[0].name]

for filename, columns in monthly_columns.items():

    if columns == reference_columns:
        print(f"{filename}: Columns match")
    else:
        print(f"{filename}: Column mismatch")
```

```python
# Verify that all 12 datasets are present
assert len(csv_files) == 12, "Expected 12 monthly CSV files."

# Load monthly datasets
dataframes = []

for file in csv_files:

    print(f"Loading {file.name}...")

    monthly_df = pd.read_csv(
        file,
        low_memory=False
    )

    dataframes.append(monthly_df)

# Combine datasets
df = pd.concat(
    dataframes,
    ignore_index=True
)

print("\nCombined dataset shape:", df.shape)

display(df.head())
```

```python
# Clean and Prepare the Dataset
# Create a working copy and remove exact duplicates

clean_df = df.copy()

original_count = len(clean_df)

clean_df = clean_df.drop_duplicates().copy()

cleaned_count = len(clean_df)

print("Original records:", original_count)
print("Records after removing duplicates:", cleaned_count)
print("Duplicates removed:", original_count - cleaned_count)
```

```python
# Verify duplicate removal

print("Duplicates in original dataset:",
      df.duplicated().sum())

print("Duplicates in cleaned dataset:",
      clean_df.duplicated().sum())
```

```python
# Convert flight dates using an explicit format

clean_df["FL_DATE"] = pd.to_datetime(
    clean_df["FL_DATE"],
    format="%m/%d/%Y %I:%M:%S %p",
    errors="coerce"
)

print(
    "Invalid dates:",
    clean_df["FL_DATE"].isna().sum()
)

print(clean_df["FL_DATE"].dtype)
```

```python
# Review Cancelled and Diverted Flights
print("Cancelled flights:")
print(
    clean_df["CANCELLED"].value_counts(
        dropna=False
    )
)

print("\nDiverted flights:")
print(
    clean_df["DIVERTED"].value_counts(
        dropna=False
    )
)
```

```python
# Recreate modeling dataset from cleaned records

model_df = clean_df[
    (clean_df["CANCELLED"] == 0) &
    (clean_df["DIVERTED"] == 0)
].copy()

model_df = model_df.dropna(
    subset=["FL_DATE", "ARR_DELAY"]
)

model_df["ARR_DEL15_TARGET"] = (
    model_df["ARR_DELAY"] >= 15
).astype(int)

print("Updated modeling dataset shape:", model_df.shape)

print("\nRemaining duplicates:")
print(model_df.duplicated().sum())
```

```python
# Create Additional Analytical Features
# Extract calendar features
model_df["FLIGHT_MONTH"] = (
    model_df["FL_DATE"].dt.month
)

model_df["FLIGHT_DAY"] = (
    model_df["FL_DATE"].dt.day
)

model_df["FLIGHT_WEEKDAY"] = (
    model_df["FL_DATE"].dt.day_of_week
)

# Extract scheduled departure hour
model_df["DEP_HOUR"] = (
    pd.to_numeric(
        model_df["CRS_DEP_TIME"],
        errors="coerce"
    ) // 100
)

# Convert 2400 to midnight
model_df["DEP_HOUR"] = (
    model_df["DEP_HOUR"].replace(24, 0)
)

# Remove invalid departure hours
model_df.loc[
    ~model_df["DEP_HOUR"].between(0, 23),
    "DEP_HOUR"
] = np.nan

display(
    model_df[
        [
            "FL_DATE",
            "FLIGHT_MONTH",
            "FLIGHT_WEEKDAY",
            "DEP_HOUR",
            "ARR_DEL15_TARGET"
        ]
    ].head()
)
```

```python
# Save the Processed Dataset
output_file = (
    processed_path /
    "flights_2025_cleaned.parquet"
)

model_df.to_parquet(
    output_file,
    index=False
)

print("Updated cleaned dataset saved successfully.")
print(output_file)
```

```python
# Compare official BTS indicator with derived target

comparison = model_df[
    ["ARR_DEL15", "ARR_DEL15_TARGET"]
].dropna()

mismatches = (
    comparison["ARR_DEL15"]
    != comparison["ARR_DEL15_TARGET"]
).sum()

print("Records compared:", len(comparison))
print("Target mismatches:", mismatches)
```
